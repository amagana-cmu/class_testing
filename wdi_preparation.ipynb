## Infrastructure Data

```python
# TODO 3: Place your solution here.

import json
import pandas as pd

# Load the nested JSON file
with open('wdi_data/2017infra.json') as f:
    data = json.load(f)

# format the data by "flattening" the nested dictionaries
reformatted_data = {}
for country, categories in data.items():
    row = {}
    # Combine stats from Transportation, Communication, and Technology
    # .get(..., {}) is a safe way to get a dictionary, or an empty one if the key is missing
    row.update(categories.get('Transportation', {}))
    row.update(categories.get('Communications', {}))
    row.update(categories.get('Technology', {}))
    reformatted_data[country] = row

# Create a DataFrame from the reformatted data
# The countries will be the index, and all 38 stats will be columns
full_infra_data = pd.DataFrame.from_dict(reformatted_data, orient='index')


# column mapping ***
infrastructure_data = full_infra_data.rename(columns={
    'Air transport, passengers carried': 'Air travel passengers',
    'Fixed broadband subscriptions (per 100 people)': 'Broadband subscriptions (per 100 people)',
    'Individuals using the Internet (% of population)': 'Internet use (% of population)',
    'High-technology exports (% of manufactured exports)': 'High-tech exports (%)',
    'Research and development expenditure (% of GDP)': 'R&D expenditure (% of GDP)',
    'Technicians in R&D (per million people)': 'Technicians in R&D (per million people)'
})

# Create a list of the columns we need to keep
cols_to_keep = [
    'Air travel passengers',
    'Broadband subscriptions (per 100 people)',
    'Internet use (% of population)',
    'High-tech exports (%)',
    'R&D expenditure (% of GDP)',
    'Technicians in R&D (per million people)'
]

# 5. Select only those 6 columns for the final DataFrame
infrastructure_data = infrastructure_data[cols_to_keep]


infrastructure_data.head()
```

## Poverty Data

```python
# TODO 4: Place your solution here.

import xml.etree.ElementTree as ET
import pandas as pd

xml_file = 'wdi_data/World_Development_Indicators_Poverty_2017.xml'

tree = ET.parse(xml_file)
root = tree.getroot()

stats_to_find = [
    'Gini index (World Bank estimate)', 
    'Income share held by highest 10%', 
    'Survey mean consumption or income per capita, total population (2011 PPP $ per day)'
]

poverty_data_list = []

# Loop through each <country> tag in the file
for country_element in root.findall('country'):
    country_name = country_element.get('name')
    stats = {'Country': country_name}
    
    # Now, loop through all <field> tags *inside* this country
    for field in country_element.findall('field'):
        metric = field.get('metric')
        if metric in stats_to_find:
            stats[metric] = field.text
            
    poverty_data_list.append(stats)

raw_poverty_data = pd.DataFrame(poverty_data_list)
raw_poverty_data = raw_poverty_data.set_index('Country')

# 3. *** RENAME THE COLUMNS ***
#    Map the raw names (from debug) to the clean names (from instructions)
poverty_data = raw_poverty_data.rename(columns={
    'Gini index (World Bank estimate)': 'Gini index',
    'Income share held by highest 10%': 'Income share of highest 10%',
    'Survey mean consumption or income per capita, total population (2011 PPP $ per day)': 'Daily consumption per capita (USD)'
})

# 4. (Optional) Select only the clean columns, just to be safe
#    This ensures the final DataFrame has only these three columns
cols_to_keep = [
    'Gini index',
    'Income share of highest 10%',
    'Daily consumption per capita (USD)'
]
poverty_data = poverty_data.astype(float)

# (Optional) Check your work
poverty_data.head()
```
