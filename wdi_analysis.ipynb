```python
# place any imports here
import pandas as pd
import os
```

## Usability Analysis

```python
# TODO 1: Place your solution here.

wdi_data = pd.read_pickle('wdi_exports/wdi_data.pkl')

country_nan_counts = wdi_data.isnull().sum(axis=1)

bins = [-1, 3, 6, 10, 15, float('inf')]

labels = ['Excellent', 'Good', 'Fair', 'Marginal', 'None']

assessment = pd.cut(country_nan_counts, bins=bins, labels=labels, right=True)


# Create a new DataFrame from our two Series
report_df = pd.DataFrame({
    'Missing Values': country_nan_counts,
    'Usability': assessment
})

report_df = report_df.sort_index()

report_df = report_df.reset_index()


report_df = report_df.rename(columns={report_df.columns[0]: 'Country Name'})


os.makedirs('wdi_exports', exist_ok=True)

report_df.to_excel('wdi_exports/country-centric-assessment.xlsx', index=False, sheet_name='Countries Missing Data')

report_df.head()
```

```python
# TODO 2: Place your solution here.

statistic_nan_counts = wdi_data.isnull().sum(axis=0)


# Bins: (..-0.1, 10], (10, 25], (25, 50], (50, 100], (100, inf]
bins = [-1, 30, 60, 100, 180, float('inf')]

# Define the labels for each group 
labels = ['Excellent', 'Good', 'Fair', 'Marginal', 'None']


# pd.cut to sort each statistic's NaN count into a category
assessment = pd.cut(statistic_nan_counts, bins=bins, labels=labels, right=True)

# Create a new DataFrame from our two Series
report_df = pd.DataFrame({
    'Missing Values': statistic_nan_counts,
    'Usability': assessment
})

# Reset the index so 'Statistic' becomes a regular column
report_df = report_df.reset_index()

# Rename the new index column to 'Statistic'
report_df = report_df.rename(columns={report_df.columns[0]: 'Statistic Name'})

# Ensure the 'wdi_exports' directory exists
os.makedirs('wdi_exports', exist_ok=True)

# Save the final report to the specified Excel file, without the pandas index [cite: 11]
report_df.to_excel('wdi_exports/statistic-centric-assessment.xlsx', index=False, sheet_name='Statistics Missing Data')

# (Optional) Check your work
report_df.head()
```

## Descriptive Statistics

```python
# TODO 3: Place your solution here.

wdi_data.index.name = 'Country'

# Define the path for our new Excel file
excel_path = 'wdi_exports/wdi_summary.xlsx'
os.makedirs('wdi_exports', exist_ok=True)


with pd.ExcelWriter(excel_path) as writer:
    
    # Loop through statistic columns
    for column_name in wdi_data.columns:
        
        #  Summary Stats ---
        max_val = wdi_data[column_name].max()
        mean_val = wdi_data[column_name].mean()
        min_val = wdi_data[column_name].min()
        
        # Create DataFrame
        summary_df = pd.DataFrame(
            {'Statistic': [max_val, mean_val, min_val]},
            index=['Max', 'Mean', 'Min']
        )
        
        # Get Top 10 ---
        # Sort by statistic (descending), then by Country (ascending) to break ties
        top_10 = wdi_data.sort_values(
            by=[column_name, 'Country'], 
            ascending=[False, True]
        )
        
        # Get the first 10 country names
        top_10_countries = top_10.index.to_list()[:10]
        
        # Create Top 10 DataFrame
        top_10_df = pd.DataFrame(
            {'Top 10': top_10_countries},
            index=pd.RangeIndex(1, 11, name='Rank')
        )
        
        # Get Bottom 10 ---
        # Sort by statistic (ascending), then by Country (ascending) to break ties
        bottom_10 = wdi_data.sort_values(
            by=[column_name, 'Country'], 
            ascending=[True, True]
        )
        
        # Get the first 10 country names
        bottom_10_countries = bottom_10.index.to_list()[:10]
        
        # Create the Bottom 10 DataFrame
        bottom_10_df = pd.DataFrame(
            {'Bottom 10': bottom_10_countries},
            index=pd.RangeIndex(1, 11, name='Rank')
        )
        
        # Write to Excel ---
        # Get sheet name (max 31 chars)
        sheet_name = column_name[:31]
        
        # Write each DataFrame to the same sheet, but in different locations
        summary_df.to_excel(writer, sheet_name=sheet_name, startrow=0, startcol=0)
        top_10_df.to_excel(writer, sheet_name=sheet_name, startrow=5, startcol=0)
        bottom_10_df.to_excel(writer, sheet_name=sheet_name, startrow=5, startcol=2)
```
